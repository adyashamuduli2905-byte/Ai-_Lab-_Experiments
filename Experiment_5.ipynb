````python
def minimax(depth,nodeIndex,isMax,scores,height):
  if depth==height:
    return scores[nodeIndex]
  if isMax:
    return max(
        minimax(depth+1,nodeIndex*2,False ,scores,height),
        minimax(depth+1,nodeIndex*2+1,False ,scores,height)
    )
  else:
    return min(
        minimax(depth+1,nodeIndex*2,True ,scores,height),
        minimax(depth+1,nodeIndex*2+1,True ,scores,height)
    )
scores=list(map(int,input("enter 8 leaf node values:").split()))
height=result=minimax(0,0,True,scores,height)
print("\n The optimal value is:",result)
````

### Explanation of the `minimax` function:

This function is the core of the algorithm and operates recursively:

-   **`def minimax(depth, nodeIndex, isMax, scores, height):`**
    -   `depth`: Represents the current level (depth) in the game tree. The root node is at `depth=0`.
    -   `nodeIndex`: Indicates the index of the current node within its level. In a binary tree, `nodeIndex * 2` and `nodeIndex * 2 + 1` are typically used to find children.
    -   `isMax`: A boolean flag that is `True` if it's the maximizing player's turn, and `False` if it's the minimizing player's turn.
    -   `scores`: A list or array containing the utility values (scores) of the leaf nodes (the final outcomes of the game).
    -   `height`: The maximum depth of the game tree, which is the depth at which leaf nodes are found.

-   **`if depth == height:`**
    -   This is the **base case** for the recursion. If the current `depth` equals the `height` of the tree, it means we have reached a leaf node. The function then returns the score directly from the `scores` list at the `nodeIndex`.

-   **`if isMax:`**
    -   If `isMax` is `True`, it's the maximizing player's turn. This player wants to choose the move that yields the highest possible score.
    -   The function recursively calls itself for the two child nodes: `minimax(depth + 1, nodeIndex * 2, False, scores, height)` and `minimax(depth + 1, nodeIndex * 2 + 1, False, scores, height)`. Notice `isMax` becomes `False` for the next level, as it will be the minimizing player's turn.
    -   `max(...)` is used to select the higher score from the two recursive calls, representing the best outcome for the maximizing player.

-   **`else:`**
    -   If `isMax` is `False`, it's the minimizing player's turn. This player wants to choose the move that yields the lowest possible score (assuming the opponent tries to maximize their own score, which would minimize the current player's score).
    -   Similar to the maximizing player, it makes recursive calls for its children, but `isMax` becomes `True` for the next level.
    -   `min(...)` is used to select the lower score from the two recursive calls, representing the best outcome for the minimizing player.
